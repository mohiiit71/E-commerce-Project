```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
pio.templates.default = "plotly_white"
```

```python
#conveting ORDER DATE &  SHIP DATE into datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
df['Order Month'] = df['Order Date'].dt.month
df['Order Year'] = df['Order Date'].dt.year
df['Order Day Of Week'] = df['Order Date'].dt.day_of_week
```

```python
df.head()
```

#Monthly Sales Analysis

```python
monthly_sales = df.groupby('Order Month')['Sales'].sum().reset_index()
fig= px.line(monthly_sales,x='Order Month',y='Sales',title='Monthly sales')
fig.show()
```

# Sales By Category

```python
sales_by_category = df.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
fig = px.pie(sales_by_category,values='Sales',names='Category',hole=0.5)
fig.show()
```

# **Sales Analysis By Sub Category**

```python
sales_by_subcategory = df.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_subcategory
```

```python
fig = px.bar(sales_by_subcategory,x='Sub-Category',y='Sales',title = "Analysis By Sub-Category")
fig.show()
```

# **Monthly Profit Analysis**

```python
profit_acc_month = df.groupby('Order Month')['Profit'].sum().reset_index()
profit_acc_month
```

# **Profit By Category**

```python
profit_by_category = df.groupby('Category')['Profit'].sum().reset_index()
profit_by_category
```

```python
fig = px.pie(profit_by_category,values="Profit",names="Category",hole=0.6)
fig.show()
```

# **Profit By Sub-Category**

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum().reset_index()
profit_by_subcategory
```

```python
fig = px.bar(profit_by_subcategory,x="Sub-Category",y="Profit")
fig.show()
```

# **SALE & PROFIT By Customer Segment**

```python
df.head()
```

```python
sales_profit_by_segment = df.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
```

```python
fig=go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],y=sales_profit_by_segment['Profit'],name="Profit"))
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],y=sales_profit_by_segment['Sales'],name="Sales"))
fig.show()
```

# Sales To Profit **Ratio**

```python
sales_profit_by_segment = df.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales']/ sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','Sales_to_Profit_Ratio']])
```
